```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')
```

```python
# Load the data
df = pd.read_csv('src/data/count_data.csv')
df.head(3)
# Change the column names
df.columns = ['Año-mes', 'año', 'mes', 'Total de puntos', 'Total de puntos sin duplicados']
```

```python
# Plot the data
df.plot(kind='bar',
        stacked=False, color=['gray', 'red'],
        x='Año-mes', y=['Total de puntos', 'Total de puntos sin duplicados'], figsize=(15, 10))

# Add labels
plt.xlabel('Año-mes')
plt.ylabel('Cantidad de puntos')
plt.title('Distribución mensual mensual de puntos de presencia con y sin duplicados')
plt.savefig('images/filtering_duplicates.png')
plt.show()
```
